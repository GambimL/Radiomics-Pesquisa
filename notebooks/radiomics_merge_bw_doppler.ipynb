# Merge de Features Radiômicas BW + Doppler

Combina os CSVs `radiomics_bw.csv` e `radiomics_doppler.csv` em um único dataset `radiomics_bw_doppler.csv`.

O merge é feito por `image_id` + `split`. Features de BW recebem prefixo `bw__` e features de Doppler recebem prefixo `doppler__`, preservando colunas de metadados (`label`, `target`, `split`) sem duplicação.

## Env Configuration

Importa bibliotecas e define os caminhos dos arquivos de entrada e saída.

```python
import sys
from pathlib import Path

import pandas as pd

MODULES_DIR = str((Path.cwd() / ".." / "modules").resolve())
if MODULES_DIR not in sys.path:
    sys.path.insert(0, MODULES_DIR)

from radiomics_utils import find_project_root
from data_merge import add_modality_prefix, merge_modalities

RADIOMICS_DIRNAME = "datasets"
INPUT_BW = "radiomics_bw.csv"
INPUT_DOPPLER = "radiomics_doppler.csv"
OUTPUT_MERGED = "radiomics_bw_doppler.csv"

META_COLS = ["image_id", "label", "target", "split"]
```

## Functions

`find_project_root` vem de `modules/radiomics_utils.py`; `add_modality_prefix` e `merge_modalities` vêm de `modules/data_merge.py`.

## Process

Carrega os dois CSVs, executa o merge com prefixo de modalidade e salva o resultado em `data/images/datasets/radiomics_bw_doppler.csv`.

```python
project_root = find_project_root(Path.cwd())
radiomics_dir = project_root / "data" / "images" / RADIOMICS_DIRNAME

bw = pd.read_csv(radiomics_dir / INPUT_BW)
doppler = pd.read_csv(radiomics_dir / INPUT_DOPPLER)

print(f"BW:     {bw.shape}")
print(f"Doppler: {doppler.shape}")
bw[META_COLS].head()
```

```python
merged = merge_modalities(bw, doppler, META_COLS)

output_path = radiomics_dir / OUTPUT_MERGED
merged.to_csv(output_path, index=False)

print(f"Salvo em: {output_path}")
print(f"Shape:    {merged.shape}")
```

```python
print("Distribuição por split e label:")
print(merged[["split", "label", "target"]].value_counts().sort_index().to_string())
print()
print("Primeiras colunas:", list(merged.columns[:8]))
print("Colunas BW (amostra):", [c for c in merged.columns if c.startswith("bw__")][:4])
print("Colunas Doppler (amostra):", [c for c in merged.columns if c.startswith("doppler__")][:4])
```
